````python
%%html
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>Gaydar v2.0 &middot; 2-layer Gemini</title>
<style>
:root{
  --host-bg:#0e1014;
  --host-fg:#e9ecf1;
  --host-fg-muted:#98a0ad;
  --host-card:#161a21;
  --host-card-fg:#e9ecf1;
  --host-input:#1d222b;
  --host-muted:#232935;
  --host-border:#2b323d;
  --host-border-hover:#3d4655;
  --host-primary:#a855f7;
  --host-secondary:#232935;
  --host-secondary-fg:#e9ecf1;
  --host-success:#22c55e;
  --host-error:#ef4444;
}
@media (prefers-color-scheme: light){
  :root{
    --host-bg:#ffffff;
    --host-fg:#111827;
    --host-fg-muted:#6b7280;
    --host-card:#f9fafb;
    --host-card-fg:#111827;
    --host-input:#ffffff;
    --host-muted:#f3f4f6;
    --host-border:#e5e7eb;
    --host-border-hover:#d1d5db;
    --host-secondary:#f3f4f6;
    --host-secondary-fg:#111827;
  }
}
*{box-sizing:border-box}
html,body{margin:0;padding:0}
body{
  background:var(--host-bg);
  color:var(--host-fg);
  font-family:ui-sans-serif,system-ui,-apple-system,"Segoe UI",Roboto,sans-serif;
  font-size:14px;line-height:1.55;-webkit-font-smoothing:antialiased;
}
.gd-root{max-width:700px;margin:0 auto;padding:28px 18px 60px}
.gd-title{font-size:26px;font-weight:500;letter-spacing:-.02em;margin:0}
.gd-title .v{font-size:11px;font-weight:500;color:var(--host-fg-muted);border:1px solid var(--host-border);border-radius:999px;padding:2px 8px;vertical-align:middle;margin-left:8px}
.gd-sub{color:var(--host-fg-muted);font-size:13px;margin:6px 0 0}
.gd-card{background:var(--host-card);border:1px solid var(--host-border);border-radius:12px;padding:16px;margin-top:14px}
.gd-card h3{margin:0 0 12px;font-size:13px;font-weight:500;letter-spacing:.02em;color:var(--host-fg-muted);text-transform:uppercase}
.gd-row{display:flex;gap:10px;align-items:center;flex-wrap:wrap}
.gd-row.tight{gap:8px}
.gd-col{display:flex;flex-direction:column;gap:6px;flex:1;min-width:200px}
label.gd-lbl{font-size:12px;color:var(--host-fg-muted)}
input[type=text],input[type=password],select,textarea{
  width:100%;background:var(--host-input);color:var(--host-fg);
  border:1px solid var(--host-border);border-radius:8px;padding:9px 11px;
  font-size:13px;font-family:inherit;outline:none}
input:focus,select:focus,textarea:focus{border-color:var(--host-border-hover)}
textarea{min-height:110px;resize:vertical;font-family:ui-monospace,SFMono-Regular,Menlo,monospace;font-size:12px}
button{background:var(--host-primary);color:#fff;border:0;border-radius:8px;padding:9px 16px;font-size:13px;font-weight:500;font-family:inherit;cursor:pointer}
button:hover{opacity:.9}
button:disabled{opacity:.45;cursor:not-allowed}
button.sec{background:var(--host-secondary);color:var(--host-secondary-fg);border:1px solid var(--host-border)}
button.sm{padding:6px 11px;font-size:12px}
.gd-drop{border:1.5px dashed var(--host-border);border-radius:12px;padding:26px 16px;text-align:center;cursor:pointer;transition:border-color .15s}
.gd-drop:hover,.gd-drop.over{border-color:#a855f7}
.gd-drop .big{font-size:15px;font-weight:500}
.gd-drop .small{font-size:12px;color:var(--host-fg-muted);margin-top:4px}
.gd-hidden{display:none!important}
.gd-pill{display:inline-flex;align-items:center;gap:6px;font-size:11px;padding:3px 9px;border-radius:999px;border:1px solid var(--host-border);color:var(--host-fg-muted)}
.gd-pill .dot{width:7px;height:7px;border-radius:50%;background:var(--host-fg-muted)}
.gd-pill.done{border-color:#22c55e;color:#22c55e}
.gd-pill.done .dot{background:#22c55e}
.gd-pill.run{border-color:#f59e0b;color:#f59e0b}
.gd-pill.run .dot{background:#f59e0b;animation:gdp 1s infinite}
.gd-pill.err{border-color:#ef4444;color:#ef4444}
.gd-pill.err .dot{background:#ef4444}
@keyframes gdp{0%,100%{opacity:1}50%{opacity:.25}}
.gd-verdict{display:flex;gap:20px;align-items:center;flex-wrap:wrap}
.gd-ring text{fill:var(--host-fg);font-family:inherit}
.gd-vlabel{font-size:17px;font-weight:500;margin-top:2px}
.gd-headline{color:var(--host-fg-muted);font-size:13px;margin-top:6px;max-width:380px}
.gd-bar{height:8px;border-radius:99px;background:var(--host-muted);overflow:hidden}
.gd-bar > i{display:block;height:100%;border-radius:99px;background:#a855f7;transition:width .6s ease}
.gd-sig{display:grid;grid-template-columns:1fr;gap:14px}
.gd-sigrow .top{display:flex;justify-content:space-between;gap:10px;font-size:13px}
.gd-sigrow .ev{font-size:12px;color:var(--host-fg-muted);margin-top:5px;padding-left:9px;border-left:2px solid var(--host-border);font-style:italic}
.gd-cats{display:grid;grid-template-columns:1fr 1fr;gap:10px 18px}
.gd-cat .lbl{display:flex;justify-content:space-between;font-size:12px;margin-bottom:4px}
.gd-quote{font-size:13px;padding:11px 13px;border-radius:8px;background:var(--host-muted);border-left:3px solid #ec4899}
.gd-err{font-size:12px;color:#ef4444;background:rgba(239,68,68,.08);border:1px solid rgba(239,68,68,.3);border-radius:8px;padding:9px 11px;white-space:pre-wrap;word-break:break-word}
.gd-note{font-size:12px;color:var(--host-fg-muted)}
.gd-cert{border:1px dashed var(--host-border);border-radius:12px;padding:18px;text-align:center}
.gd-cert .t{font-size:11px;letter-spacing:.22em;color:var(--host-fg-muted)}
.gd-cert .n{font-size:20px;font-weight:500;margin:8px 0 4px}
.gd-cert .c{font-size:12px;color:var(--host-fg-muted)}
.gd-cert .s{font-size:11px;color:var(--host-fg-muted);margin-top:10px;font-family:ui-monospace,monospace}
.gd-foot{font-size:11px;color:var(--host-fg-muted);margin-top:16px;text-align:center}
.gd-legend{display:flex;flex-wrap:wrap;gap:14px;font-size:11px;color:var(--host-fg-muted);margin-bottom:6px}
.gd-legend span{display:flex;align-items:center;gap:5px}
.gd-legend i{width:9px;height:9px;border-radius:2px;display:block}
@media(max-width:520px){.gd-cats{grid-template-columns:1fr}}
</style>
</head>
<body>
<div class="gd-root" id="gdRoot">
  <div>
    <h1 class="gd-title">Gaydar<span class="v">v2.0 &middot; 2-layer gemini</span></h1>
    <p class="gd-sub">Drop in a WhatsApp export. Layer 0 profiles the raw stats locally, Layer 1 extracts behavioural signals with Gemini, Layer 2 cross-examines the evidence and returns a verdict.</p>
  </div>

  <div class="gd-card">
    <h3>1 &middot; engine</h3>
    <div class="gd-row">
      <div class="gd-col">
        <label class="gd-lbl">Gemini API key</label>
        <input type="password" id="gdKey" placeholder="AIza..." autocomplete="off">
      </div>
      <div class="gd-col">
        <label class="gd-lbl">Model</label>
        <select id="gdModel">
          <option value="gemini-3.5-flash-lite" selected>gemini-3.5-flash-lite</option>
          <option value="gemini-2.5-flash-lite">gemini-2.5-flash-lite</option>
          <option value="gemini-2.5-flash">gemini-2.5-flash</option>
          <option value="gemini-2.5-pro">gemini-2.5-pro</option>
          <option value="gemini-2.0-flash">gemini-2.0-flash</option>
        </select>
      </div>
    </div>
    <div class="gd-row" style="margin-top:10px">
      <button class="sec sm" id="gdFetchModels">Fetch my available models</button>
      <button class="sec sm" id="gdSaveKey">Save key</button>
      <span class="gd-note" id="gdKeyState">key stored in this browser only</span>
    </div>
  </div>

  <div class="gd-card">
    <h3>2 &middot; chat history</h3>
    <div class="gd-drop" id="gdDrop">
      <div class="big">Drop a WhatsApp .txt or .zip export here</div>
      <div class="small">or click to choose a file &middot; nothing leaves your device except the sampled stats</div>
      <input type="file" id="gdFile" accept=".txt,.zip,text/plain,application/zip" class="gd-hidden">
    </div>
    <div class="gd-row" style="margin-top:10px">
      <button class="sec sm" id="gdDemo">Load demo chat</button>
      <button class="sec sm" id="gdPasteToggle">Paste text instead</button>
      <span class="gd-note" id="gdFileState"></span>
    </div>
    <div id="gdPasteWrap" class="gd-hidden" style="margin-top:10px">
      <textarea id="gdPaste" placeholder="[12/03/2026, 21:07:14] Sam: yesss finally 🙌"></textarea>
      <button class="sm" id="gdParsePaste" style="margin-top:8px">Parse pasted text</button>
    </div>

    <div id="gdTargetWrap" class="gd-hidden" style="margin-top:14px">
      <div class="gd-row">
        <div class="gd-col">
          <label class="gd-lbl">Who are we scanning?</label>
          <select id="gdTarget"></select>
        </div>
        <div class="gd-col" style="flex:0 0 auto;min-width:0">
          <label class="gd-lbl">&nbsp;</label>
          <button id="gdRun">Run the gaydar</button>
        </div>
      </div>
      <div class="gd-note" id="gdChatState" style="margin-top:8px"></div>
    </div>
  </div>

  <div class="gd-card gd-hidden" id="gdPipeCard">
    <h3>pipeline</h3>
    <div class="gd-row tight" id="gdPipe"></div>
    <div class="gd-err gd-hidden" id="gdPipeErr" style="margin-top:10px"></div>
  </div>

  <div id="gdResults"></div>

  <p class="gd-foot">Gaydar is vibes-based, not science. It reads style, not biology. Layer 0 runs entirely offline in your browser.</p>
</div>

<script src="https://cdnjs.cloudflare.com/ajax/libs/Chart.js/4.4.1/chart.umd.js"></script>
<script>
(function(){
"use strict";

/* ============================== utils ============================== */
var $ = function(s,r){ return (r||document).querySelector(s); };
var clamp = function(v,a,b){ return Math.max(a,Math.min(b,v)); };
var esc = function(s){ return String(s==null?"":s).replace(/[&<>"']/g,function(c){return {"&":"&amp;","<":"&lt;",">":"&gt;",'"':"&quot;","'":"&#39;"}[c];}); };
var store = {
  get:function(k,d){ try{ var v=localStorage.getItem("gaydar_"+k); return v==null?d:JSON.parse(v); }catch(e){ return d; } },
  set:function(k,v){ try{ localStorage.setItem("gaydar_"+k, JSON.stringify(v)); }catch(e){} }
};
var state = { chat:null, target:null, local:null, l1:null, l2:null, running:false, chart:null };

/* ============================== whatsapp parser ============================== */
var TS_RE = /^\u200e?\[?(\d{1,4}[\/.\-]\d{1,2}[\/.\-]\d{1,4}),?\s+(\d{1,2}:\d{2}(?::\d{2})?)\s*([AaPp]\.?\s?[Mm]\.?)?\]?\s*[\-\u2013\u2014]?\s*([\s\S]*)$/;
var AUTHOR_RE = /^([^:]{1,80}?):\s?([\s\S]*)$/;
var MEDIA_RE = /^(<Media omitted>|(image|video|sticker|GIF|audio|document|Contact card|Location|Voice message)\s*(omitted|attached)?$)/i;
var DEL_RE = /^(This message was deleted|You deleted this message)$/i;
var SYS_RE = /^(Messages and calls are end-to-end encrypted|You (added|removed|changed|created|joined|left|pinned|deleted|were added|were removed)|[^:]{1,60}\s(joined using this group|changed the subject|changed this group|changed the group|changed their phone|created this group|added |removed |left$|pinned a message|was added)|null)/i;

function toDate(dateStr, timeStr, ampm){
  try{
    var d = dateStr.split(/[\/.\-]/).map(Number), p = timeStr.split(":").map(Number);
    if(d.length<3) return null;
    var day, mon, yr;
    if(d[0]>31){ yr=d[0]; mon=d[1]; day=d[2]; }
    else { day=d[0]; mon=d[1]; yr=d[2]<100?(d[2]>70?1900+d[2]:2000+d[2]):d[2]; }
    var h=p[0];
    if(ampm){ var pm=/p/i.test(ampm); if(pm&&h<12)h+=12; if(!pm&&h===12)h=0; }
    return new Date(yr,mon-1,day,h,p[1]||0,p[2]||0);
  }catch(e){ return null; }
}

function parseWhatsApp(raw){
  raw = String(raw).replace(/^\uFEFF/,"").replace(/\u200e/g,"").replace(/\r\n?/g,"\n");
  var lines = raw.split("\n"), msgs=[], cur=null;
  for(var i=0;i<lines.length;i++){
    var line = lines[i], m = line.match(TS_RE);
    if(m){
      var rest = m[4] || "";
      var am = rest.match(AUTHOR_RE);
      if(!am || am[1].length>40){ msgs.push({d:toDate(m[1],m[2],m[3]),author:"__sys__",text:rest,type:"sys"}); cur=msgs[msgs.length-1]; continue; }
      var author = am[1].trim(), body = am[2] || "", type = "msg";
      if(MEDIA_RE.test(body)) type="media";
      else if(DEL_RE.test(body)) type="deleted";
      else if(SYS_RE.test(body)) type="sys";
      cur = {d:toDate(m[1],m[2],m[3]), author:author, text:body, type:type};
      msgs.push(cur);
    } else {
      if(cur && line.trim()!=="") cur.text += "\n"+line;
    }
  }
  return msgs;
}

/* ============================== layer 0: local heuristics ============================== */
var CATS = [
  {id:"emoji",     label:"emoji palette",          icon:"\uD83C\uDF08", w:1.00},
  {id:"slang",     label:"camp lexicon",           icon:"\uD83D\uDC85", w:1.30},
  {id:"drama",     label:"dramatic register",      icon:"\uD83C\uDFAD", w:0.70},
  {id:"affection", label:"affection & pet names",  icon:"\uD83D\uDC96", w:1.00},
  {id:"aesthetic", label:"aesthetic obsession",    icon:"\uD83D\uDD6F", w:1.10},
  {id:"pop",       label:"pop culture fluency",    icon:"\uD83C\uDFA4", w:1.20},
  {id:"thirst",    label:"thirst & gym",           icon:"\uD83D\uDD25", w:0.90},
  {id:"theatre",   label:"musical theatre energy", icon:"\uD83C\uDFBC", w:1.10},
  {id:"social",    label:"social butterfly",       icon:"\uD83D\uDCAC", w:0.40},
  {id:"hetero",    label:"straight-coding",        icon:"\uD83C\uDF7A", w:0.00}
];

var SIGS = [
  {cat:"emoji", n:"rainbow & sparkle emoji", re:/[\uD83C\uDF08\u2728\uD83D\uDCAB\u2B50\uD83E\uDD84\uD83E\uDEE7]/gu, rate:.06, w:3.0},
  {cat:"emoji", n:"heart emoji", re:/[\u2764\uFE0F\uD83D\uDC96\uD83D\uDC95\uD83D\uDC97\uD83D\uDC93\uD83D\uDC9E\uD83D\uDC98\uD83E\uDD70\uD83D\uDE0D\uD83D\uDE18]/gu, rate:.08, w:2.5},
  {cat:"emoji", n:"sassy hands & nails", re:/[\uD83D\uDC85\uD83D\uDE4C\uD83D\uDC4F\uD83D\uDC81\uD83D\uDE46]/gu, rate:.05, w:2.0},
  {cat:"emoji", n:"flowers & bows", re:/[\uD83C\uDF38\uD83C\uDF37\uD83C\uDF3A\uD83D\uDC90\uD83C\uDF80\uD83C\uDF39]/gu, rate:.05, w:2.5},
  {cat:"emoji", n:"fire & thirst emoji", re:/[\uD83D\uDD25\uD83E\uDD75\uD83D\uDE29\uD83D\uDC40]/gu, rate:.06, w:1.5},
  {cat:"emoji", n:"peach & eggplant", re:/[\uD83C\uDF51\uD83C\uDF46]/gu, rate:.02, w:3.0},
  {cat:"emoji", n:"cry-laugh emoji", re:/[\uD83D\uDE02\uD83E\uDD23\uD83D\uDC80\uD83D\uDE2D]/gu, rate:.15, w:1.0},
  {cat:"emoji", n:"kiss emoji", re:/[\uD83D\uDC8B\uD83D\uDC44]/gu, rate:.03, w:2.0},
  {cat:"slang", n:"core camp vocabulary", re:/\b(yass+|yasss+|slay(?:ed|ing|s)?|queen|hunty|sis|periodt?|werk|fierce|snatched|gagged|wig|spill)\b/gi, rate:.05, w:4.0},
  {cat:"slang", n:"obsessed / iconic / vibe", re:/\b(iconic|obsessed|vibe[sd]?|serving|serve[sd]?|aesthetic|era|moment|lowkey|highkey|literally|dead|dying|mood)\b/gi, rate:.10, w:2.0},
  {cat:"slang", n:"omg family", re:/\b(omg+|omfg+|omgg+)\b/gi, rate:.06, w:1.5},
  {cat:"slang", n:"pet names & praise", re:/\b(babe|babs|hun|honey|sweetie|darling|gorgeous|stunning|mwah|love you)\b/gi, rate:.06, w:3.0},
  {cat:"slang", n:"keysmash & drawn-out laughs", re:/(~~~|asdfjkl|ksksks+|lmaoo+|lolol+|hahahahaha)/gi, rate:.05, w:2.0},
  {cat:"drama", n:"multiple exclamation marks", re:/!{2,}/g, rate:.12, w:1.5},
  {cat:"drama", n:"stretched words", re:/\b\w*(\w)\1{2,}\w*\b/g, rate:.10, w:2.0},
  {cat:"drama", n:"SHOUTING in caps", re:/\b[A-Z]{3,}\b/g, rate:.08, w:1.5},
  {cat:"affection", n:"warmth & compliments", re:/\b(cute|adorable|beautiful|gorgeous|stunning|pretty|love|adore|obsessed with|so sweet|proud of you)\b/gi, rate:.12, w:2.5},
  {cat:"affection", n:"checking in / feelings talk", re:/\b(how are you|you ok|you good|miss you|thinking of you|how was|hope you|proud of)\b/gi, rate:.08, w:1.5},
  {cat:"aesthetic", n:"aesthetic vocabulary", re:/\b(aesthetic|aesthetics|aura|vibe|vibes|fit|outfit|thrift|vintage|cute|d\u00e9cor|interior|colour palette|color palette)\b/gi, rate:.10, w:2.5},
  {cat:"aesthetic", n:"homeware & candles", re:/\b(candle|throw pillow|plants?|monstera|pampas|ikea|muji|ceramics|linen|beige|sage green|scrunchie)\b/gi, rate:.05, w:2.0},
  {cat:"aesthetic", n:"skincare & self-care", re:/\b(skincare|serum|retinol|spf|routine|face mask|moisturis|moisturiz|hydrat|glow)\b/gi, rate:.04, w:2.5},
  {cat:"pop", n:"diva mentions", re:/\b(beyonc\u00e9|beyonce|taylor swift|lady gaga|gaga|ariana|rihanna|britney|mariah|whitney|cher|madonna|kylie|dua lipa|chappell|charli xcx|olivia rodrigo|sabrina carpenter|nicki|cardi b|megan thee)\b/gi, rate:.06, w:2.5},
  {cat:"pop", n:"drag & reality tv", re:/\b(drag race|rupaul|drag|reality tv|love island|too hot to handle|the bachelor|housewives|kardashian|big brother)\b/gi, rate:.05, w:3.0},
  {cat:"pop", n:"stan / fandom behaviour", re:/\b(stan|era|tour|tickets|concert|merch|vinyl|setlist|opening night)\b/gi, rate:.05, w:1.5},
  {cat:"theatre", n:"musical theatre", re:/\b(musical|musicals|broadway|west end|wicked|hamilton|les mis|rent|mamma mia|chicago|phantom|sondheim|cast recording|soundtrack|showtune)\b/gi, rate:.05, w:3.0},
  {cat:"theatre", n:"performing & drama club", re:/\b(rehearsal|audition|chorus|harmon|belt(ing)?|karaoke|open mic|improv|stage)\b/gi, rate:.04, w:2.0},
  {cat:"thirst", n:"gym & gains", re:/\b(gym|gains|reps|squat|bench|deadlift|protein|shredded|abs|bulk|cutting|leg day|pr)\b/gi, rate:.07, w:1.5},
  {cat:"thirst", n:"rating people", re:/\b(hot|handsome|fine|thirst|shirtless|beard|jawline|forearms|arms|abs|so fit)\b/gi, rate:.08, w:2.5},
  {cat:"thirst", n:"celebrity crushes", re:/\b(chris (hemsworth|evans)|henry cavill|timoth\u00e9e|timothee|harry styles|jacob elordi|pedro pascal|ryan gosling|zayn|jungkook|bts)\b/gi, rate:.03, w:3.0},
  {cat:"social", n:"voice notes & calls", re:/\b(voice note|vn|call me|facetime|ft me|ring me)\b/gi, rate:.05, w:1.5},
  {cat:"social", n:"questions back", re:/\?/g, rate:.25, w:0.5},
  {cat:"social", n:"social plans", re:/\b(brunch|dinner|drinks|party|rooftop|mimosa|plans|sunday roast)\b/gi, rate:.08, w:1.0},
  {cat:"hetero", n:"bro / bruh / mate", re:/\b(bro|bruh|dude|mate|lad|fella|my guy|chief)\b/gi, rate:.20, w:-1.5},
  {cat:"hetero", n:"the no-homo clause", re:/\bno homo\b/gi, rate:.01, w:-4.0},
  {cat:"hetero", n:"sport & pints", re:/\b(football|soccer|premier league|nfl|nba|beer|pint|barbecue|bbq|fishing|rugby)\b/gi, rate:.10, w:-1.5},
  {cat:"hetero", n:"gaming & cars", re:/\b(gaming|fifa|call of duty|warzone|pc build|gpu|motorbike|engine)\b/gi, rate:.06, w:-1.0}
];

function layer0(targetName){
  var msgs = state.chat.messages.filter(function(m){ return m.author===targetName && m.type!=="sys"; });
  var texts = msgs.filter(function(m){ return m.type==="msg"; }).map(function(m){ return m.text; });
  var n = Math.max(texts.length,1);
  var joined = texts.join("\n");
  var cats = {}, details = [], totalHits = 0;

  CATS.forEach(function(c){ cats[c.id] = {id:c.id,label:c.label,icon:c.icon,w:c.w,score:0,sum:0,abs:0}; });

  SIGS.forEach(function(s){
    var hits = joined.match(s.re);
    var count = hits ? hits.length : 0;
    if(count===0) return;
    totalHits += count;
    var threshold = Math.max(2, n * s.rate);
    var strength = clamp(count/threshold, 0, 1);
    var c = cats[s.cat];
    c.sum += strength * Math.abs(s.w);
    c.abs += Math.abs(s.w);
    details.push({cat:s.cat, name:s.n, count:count, strength:strength, weight:s.w});
  });

  CATS.forEach(function(c){
    var cc = cats[c.id];
    cc.score = cc.abs ? Math.round(clamp(cc.sum/cc.abs,0,1)*100) : 0;
  });

  var weighted=0, wsum=0;
  CATS.forEach(function(c){
    if(c.id==="hetero") return;
    weighted += cats[c.id].score * c.w;
    wsum += c.w;
  });
  var base = wsum ? weighted/wsum : 0;
  var penalty = cats.hetero.score * 0.22;

  var chars = joined.replace(/\s/g,"").length;
  var emojiCount = (joined.match(/\p{Extended_Pictographic}/gu)||[]).length;
  var lateNight = 0;
  msgs.forEach(function(m){ if(m.d){ var h=m.d.getHours(); if(h>=0&&h<4) lateNight++; } });

  var idx = Math.round(clamp(base - penalty, 0, 100));
  details.sort(function(a,b){ return (b.strength*Math.abs(b.weight)) - (a.strength*Math.abs(a.weight)); });

  var sortedCats = CATS.filter(function(c){return c.id!=="hetero";}).map(function(c){return cats[c.id];})
    .sort(function(a,b){ return b.score-a.score; });

  return {
    index: idx,
    cats: cats,
    catList: sortedCats,
    details: details.slice(0,18),
    counts: {
      messages: texts.length,
      media: msgs.filter(function(m){return m.type==="media";}).length,
      chars: chars,
      avgLen: texts.length ? Math.round(chars/texts.length) : 0,
      emoji: emojiCount,
      emojiPerMsg: texts.length ? +(emojiCount/texts.length).toFixed(2) : 0,
      lateNight: lateNight,
      totalHits: totalHits
    },
    topCat: sortedCats[0] ? sortedCats[0].label : "n/a",
    heteroScore: cats.hetero.score
  };
}

/* ============================== sample builder ============================== */
function buildTranscript(targetName, maxChars){
  maxChars = maxChars || 9000;
  var all = state.chat.messages.filter(function(m){ return m.type!=="sys" && m.text; });
  if(!all.length) return "";
  var windows = [0.08,0.33,0.58,0.83], picked = [];
  windows.forEach(function(p){
    var start = Math.max(0, Math.floor(all.length*p) - 12);
    picked = picked.concat(all.slice(start, start+26));
  });
  var seen = {}, out = [];
  for(var i=0;i<picked.length;i++){
    var m = picked[i], key = (m.d ? m.d.getTime() : i) + "|" + m.author;
    if(seen[key]) continue;
    seen[key] = 1;
    var tag = m.author===targetName ? ">>> " : "    ";
    out.push(tag + m.author + ": " + m.text.replace(/\n/g," ").slice(0,240));
  }
  var txt = out.join("\n");
  if(txt.length>maxChars) txt = txt.slice(txt.length-maxChars);
  return txt;
}

/* ============================== gemini ============================== */
var API = "https://generativelanguage.googleapis.com/v1beta/models/";

function gemini(opts){
  var url = API + encodeURIComponent(opts.model) + ":generateContent?key=" + encodeURIComponent(opts.apiKey);
  var body = {
    systemInstruction: { parts: [{ text: opts.system }] },
    contents: [{ role:"user", parts: [{ text: opts.prompt }] }],
    generationConfig: { temperature: opts.temp==null?0.7:opts.temp, responseMimeType:"application/json" }
  };
  if(opts.schema) body.generationConfig.responseSchema = opts.schema;
  return fetch(url, {
    method:"POST",
    headers:{ "Content-Type":"application/json" },
    body: JSON.stringify(body)
  }).then(function(r){
    return r.text().then(function(t){
      if(!r.ok){
        var msg = t;
        try{ var j=JSON.parse(t); if(j.error&&j.error.message) msg = j.error.message; }catch(e){}
        var err = new Error("Gemini " + r.status + ": " + String(msg).slice(0,400));
        err.status = r.status; throw err;
      }
      return JSON.parse(t);
    });
  }).then(function(j){
    var c = j.candidates && j.candidates[0];
    if(!c) throw new Error("No candidates returned. " + JSON.stringify(j).slice(0,300));
    if(c.finishReason && c.finishReason!=="STOP" && c.finishReason!=="MAX_TOKENS") throw new Error("Blocked or truncated: " + c.finishReason);
    var parts = (c.content && c.content.parts) || [];
    var txt = parts.map(function(p){ return p.text||""; }).join("");
    if(!txt) throw new Error("Empty response. " + JSON.stringify(j).slice(0,300));
    var clean = txt.trim().replace(/^```(?:json)?/i,"").replace(/```$/,"").trim();
    try { return JSON.parse(clean); } catch(e){ throw new Error("Model returned non-JSON: " + clean.slice(0,200)); }
  });
}

var SCHEMA_L1 = {
  type:"object",
  properties:{
    signals:{ type:"array", items:{ type:"object", properties:{
      name:{type:"string"}, category:{type:"string"}, weight:{type:"number"},
      evidence:{type:"string"}, note:{type:"string"}
    }, required:["name","category","weight","evidence"] } },
    topics:{ type:"array", items:{type:"string"} },
    tone:{ type:"array", items:{type:"string"} },
    catchphrases:{ type:"array", items:{type:"string"} },
    straight_coded:{ type:"array", items:{ type:"object", properties:{
      name:{type:"string"}, evidence:{type:"string"}
    }, required:["name","evidence"] } }
  },
  required:["signals","topics","tone","catchphrases","straight_coded"]
};

var SCHEMA_L2 = {
  type:"object",
  properties:{
    score:{type:"integer"}, verdict:{type:"string"}, confidence:{type:"integer"},
    headline:{type:"string"}, reasoning:{type:"string"},
    top_signals:{ type:"array", items:{ type:"object", properties:{
      name:{type:"string"}, impact:{type:"integer"}, evidence:{type:"string"}
    }, required:["name","impact","evidence"] } },
    straight_case:{type:"string"}, roast:{type:"string"}, certification:{type:"string"}
  },
  required:["score","verdict","confidence","headline","reasoning","top_signals","straight_case","roast","certification"]
};

var SYS_L1 = "You are GAYDAR-L1, a behavioural linguist who infers personality and orientation-adjacent vibes purely from chat style. You ignore facts and care only about register, vocabulary, emoji palette, rhythm and topics. You are precise but funny. Output strict JSON only.";
var SYS_L2 = "You are GAYDAR-L2, chief adjudicator of a joke-but-serious gaydar engine. You fuse a local statistical layer with L1's extracted signals and hand down a final verdict. You are decisive, funny and never hedge into meaninglessness. Output strict JSON only.";

function runLayer1(name, transcript, model, key){
  var prompt = [
    "TARGET: " + name,
    "TRANSCRIPT EXCERPTS (WhatsApp, target's own messages are marked with >>>; others are context):",
    "",
    transcript,
    "",
    "TASK: extract the strongest stylistic signals in TARGET's own messages.",
    "For each signal give: a short name, a category from [emoji, slang, drama, affection, aesthetic, pop, thirst, theatre, social, straight], a weight from -5 (strongly straight-coded) to +5 (strongly gay-coded), a verbatim evidence quote under 120 characters, and a one-line note.",
    "Give 8 to 14 signals, strongest first. Quote real text, never invent it.",
    "Also list: recurring topics, 3-5 tone adjectives, signature catchphrases (verbatim), and straight_coded items with evidence.",
    "Judge only style. Do not assume anything about the other participants."
  ].join("\n");
  return gemini({ apiKey:key, model:model, system:SYS_L1, prompt:prompt, schema:SCHEMA_L1, temp:0.6 });
}

function runLayer2(name, local, l1, meta, model, key){
  var catLines = local.catList.map(function(c){ return "  " + c.label + ": " + c.score + "/100"; }).join("\n");
  var prompt = [
    "TARGET: " + name,
    "MESSAGES ANALYSED: " + local.counts.messages + " of " + meta.totalMessages + " in the chat",
    "CHAT SPAN: " + meta.span,
    "PARTICIPANTS: " + meta.participants,
    "",
    "LAYER 0 - LOCAL STATISTICAL LAYER (computed in-browser, no AI):",
    "  local_index: " + local.index + "/100",
    "  category scores (0-100):",
    catLines,
    "  straight-coding score: " + local.heteroScore + "/100",
    "  emoji per message: " + local.counts.emojiPerMsg,
    "  avg message length: " + local.counts.avgLen + " chars",
    "  media messages: " + local.counts.media,
    "  messages sent between midnight and 4am: " + local.counts.lateNight,
    "  dominant category: " + local.topCat,
    "  top raw signal hits: " + local.details.slice(0,10).map(function(d){ return d.name + " x" + d.count; }).join(", "),
    "",
    "LAYER 1 - GEMINI SIGNAL EXTRACTION (JSON):",
    JSON.stringify(l1, null, 1).slice(0,6000),
    "",
    "TASK: issue the final verdict.",
    "- score: integer 0-100 gayness. Anchor it: 50 is a coin flip, 90 is a total eclipse.",
    "- verdict: a short punchy label, max 5 words.",
    "- confidence: integer 0-100 based on how much material there was.",
    "- headline: one sentence, max 140 chars.",
    "- reasoning: 2 to 4 sentences that cite actual evidence from the transcript.",
    "- top_signals: 3 to 5 items with impact -5..5 and verbatim evidence.",
    "- straight_case: 1-2 sentences giving the strongest argument for straight.",
    "- roast: one savage but affectionate line about the target.",
    "- certification: one mock-official sentence, e.g. 'Certified 78% gay by the Gaydar engine, layer 2 of 2.'",
    "Rules: this is vibes, not science. Weigh Layer 1 against Layer 0; if they disagree, say which you trust and why. Thin evidence lowers confidence, not the score."
  ].join("\n");
  return gemini({ apiKey:key, model:model, system:SYS_L2, prompt:prompt, schema:SCHEMA_L2, temp:0.8 });
}

/* ============================== verdict bands ============================== */
function band(s){
  if(s<12) return "certified hetero (allegedly)";
  if(s<28) return "straight-passing, with a skincare routine";
  if(s<45) return "suspicious";
  if(s<60) return "schr\u00f6dinger's sexuality";
  if(s<75) return "loudly not-straight";
  if(s<88) return "flaming, officially";
  return "total gay eclipse";
}

/* ============================== rendering ============================== */
function pipeStep(id, label, status, note){
  var el = document.getElementById("gdP_"+id);
  if(!el){
    el = document.createElement("span");
    el.id = "gdP_"+id;
    el.className = "gd-pill";
    el.innerHTML = '<span class="dot"></span><span class="txt"></span>';
    $("#gdPipe").appendChild(el);
  }
  el.className = "gd-pill " + (status==="done"?"done":status==="run"?"run":status==="err"?"err":"");
  el.querySelector(".txt").textContent = label + (note ? " \u00b7 " + note : "");
}
function pipeErr(msg){
  var e = $("#gdPipeErr");
  if(!msg){ e.classList.add("gd-hidden"); return; }
  e.textContent = msg; e.classList.remove("gd-hidden");
}

function ring(pct, color){
  var C = 2*Math.PI*80, off = C - (C*pct/100);
  return '<svg class="gd-ring" viewBox="0 0 200 200" width="168" height="168" aria-hidden="true">' +
    '<circle cx="100" cy="100" r="80" fill="none" stroke="var(--host-border)" stroke-width="13"/>' +
    '<circle cx="100" cy="100" r="80" fill="none" stroke="'+color+'" stroke-width="13" stroke-linecap="round" ' +
    'stroke-dasharray="'+C.toFixed(1)+'" stroke-dashoffset="'+off.toFixed(1)+'" transform="rotate(-90 100 100)"/>' +
    '<text x="100" y="95" text-anchor="middle" font-size="40" font-weight="500">'+pct+'%</text>' +
    '<text x="100" y="118" text-anchor="middle" font-size="11" fill="var(--host-fg-muted)">gaydar index</text>' +
    '</svg>';
}

function render(){
  var box = $("#gdResults");
  var local = state.local;
  if(!local){ box.innerHTML=""; return; }
  var name = state.target;
  var l2 = state.l2, l1 = state.l1;
  var score = l2 ? clamp(Math.round(l2.score),0,100) : local.index;
  var color = score>=75 ? "#ec4899" : score>=50 ? "#a855f7" : score>=28 ? "#22d3ee" : "#22c55e";
  var verdict = l2 && l2.verdict ? l2.verdict : band(score);
  var conf = l2 && l2.confidence!=null ? clamp(Math.round(l2.confidence),0,100) : (l1?45:25);
  var headline = l2 && l2.headline ? l2.headline : "Local statistics only \u2014 add a Gemini key to run layers 1 and 2.";
  var serial = "GD-" + Math.abs(hashStr(name + state.chat.meta.span + score)).toString(36).slice(0,6).toUpperCase() + "-" + score;

  var h = "";
  h += '<div class="gd-card">';
  h += '<h3>verdict &middot; ' + esc(name) + '</h3>';
  h += '<div class="gd-verdict">' + ring(score, color);
  h += '<div style="flex:1;min-width:220px">';
  h += '<div class="gd-vlabel">' + esc(verdict) + '</div>';
  h += '<div class="gd-note" style="margin-top:4px">confidence ' + conf + '% &middot; ' + (l2 ? "layers 0-2 complete" : "layer 0 only") + '</div>';
  h += '<div class="gd-headline">' + esc(headline) + '</div>';
  if(l2 && l2.roast){ h += '<div class="gd-quote" style="margin-top:12px">\u201c' + esc(l2.roast) + '\u201d</div>'; }
  h += '</div></div>';
  if(l2 && l2.reasoning){ h += '<p style="margin:16px 0 0;font-size:13px">' + esc(l2.reasoning) + '</p>'; }
  h += '</div>';

  if(l2 && l2.top_signals && l2.top_signals.length){
    h += '<div class="gd-card"><h3>the case against them</h3><div class="gd-sig">';
    l2.top_signals.forEach(function(s){
      var imp = clamp(Number(s.impact)||0,-5,5);
      var pct = Math.round(Math.abs(imp)/5*100);
      var col = imp>0 ? "#ec4899" : "#22c55e";
      h += '<div class="gd-sigrow"><div class="top"><span>' + esc(s.name) + '</span><span style="color:'+col+'">' + (imp>0?"+":"") + imp + '</span></div>';
      h += '<div class="gd-bar" style="margin-top:5px"><i style="width:'+pct+'%;background:'+col+'"></i></div>';
      if(s.evidence) h += '<div class="ev">' + esc(s.evidence) + '</div>';
      h += '</div>';
    });
    h += '</div></div>';
  }

  h += '<div class="gd-card"><h3>local vibe profile &middot; layer 0</h3>';
  h += '<div class="gd-legend"><span><i style="background:#a855f7"></i>gay-coded category</span><span><i style="background:#22c55e"></i>straight-coded category</span></div>';
  h += '<div style="position:relative;width:100%;height:300px"><canvas id="gdRadar"></canvas></div>';
  h += '<div class="gd-cats" style="margin-top:14px">';
  local.catList.forEach(function(c){
    h += '<div class="gd-cat"><div class="lbl"><span>' + c.icon + ' ' + esc(c.label) + '</span><span style="color:var(--host-fg-muted)">' + c.score + '</span></div>';
    h += '<div class="gd-bar"><i style="width:'+c.score+'%"></i></div></div>';
  });
  h += '<div class="gd-cat" style="margin-top:12px"><div class="lbl"><span>\uD83C\uDF7A straight-coding</span><span style="color:var(--host-fg-muted)">' + local.heteroScore + '</span></div>';
  h += '<div class="gd-bar"><i style="width:'+local.heteroScore+'%;background:#22c55e"></i></div></div>';
  h += '</div>';
  h += '<div class="gd-row" style="margin-top:16px;gap:18px">';
  h += '<span class="gd-note">' + local.counts.messages + ' messages analysed</span>';
  h += '<span class="gd-note">' + local.counts.emojiPerMsg + ' emoji/msg</span>';
  h += '<span class="gd-note">' + local.counts.avgLen + ' chars avg</span>';
  h += '<span class="gd-note">' + local.counts.lateNight + ' late-night msgs</span>';
  h += '</div></div>';

  if(l1){
    h += '<div class="gd-card"><h3>layer 1 &middot; extracted signals</h3><div class="gd-sig">';
    (l1.signals||[]).slice(0,12).forEach(function(s){
      var w = clamp(Number(s.weight)||0,-5,5);
      var pct = Math.round(Math.abs(w)/5*100);
      var col = w>0 ? "#a855f7" : "#22c55e";
      h += '<div class="gd-sigrow"><div class="top"><span>' + esc(s.name) + ' <span class="gd-note">(' + esc(s.category) + ')</span></span><span style="color:'+col+'">' + (w>0?"+":"") + w + '</span></div>';
      h += '<div class="gd-bar" style="margin-top:5px"><i style="width:'+pct+'%;background:'+col+'"></i></div>';
      if(s.evidence) h += '<div class="ev">' + esc(s.evidence) + '</div>';
      if(s.note) h += '<div class="gd-note" style="margin-top:4px">' + esc(s.note) + '</div>';
      h += '</div>';
    });
    h += '</div>';
    if(l1.tone && l1.tone.length){ h += '<div class="gd-row" style="margin-top:14px">' + l1.tone.map(function(t){ return '<span class="gd-pill">'+esc(t)+'</span>'; }).join("") + '</div>'; }
    if(l1.catchphrases && l1.catchphrases.length){ h += '<p class="gd-note" style="margin-top:12px">catchphrases: ' + l1.catchphrases.map(esc).join(" \u00b7 ") + '</p>'; }
    if(l1.topics && l1.topics.length){ h += '<p class="gd-note" style="margin-top:6px">recurring topics: ' + l1.topics.map(esc).join(", ") + '</p>'; }
    h += '</div>';
  }

  if(l2 && (l2.straight_case || (l1 && l1.straight_coded && l1.straight_coded.length))){
    h += '<div class="gd-card"><h3>the straight case</h3>';
    if(l2.straight_case) h += '<p style="margin:0;font-size:13px">' + esc(l2.straight_case) + '</p>';
    ((l1 && l1.straight_coded) || []).slice(0,4).forEach(function(s){
      h += '<div class="gd-sigrow" style="margin-top:10px"><div class="top"><span>' + esc(s.name) + '</span></div><div class="ev">' + esc(s.evidence) + '</div></div>';
    });
    h += '</div>';
  }

  h += '<div class="gd-card gd-cert"><div class="t">CERTIFICATE OF SEXUAL ORIENTATION VIBES</div>';
  h += '<div class="n">' + esc(name) + '</div>';
  h += '<div class="c">' + (l2 && l2.certification ? esc(l2.certification) : 'Certified ' + score + '% gay by the Gaydar engine, layer 0 of 2.') + '</div>';
  h += '<div class="s">serial ' + serial + ' &middot; ' + esc(state.chat.meta.span) + ' &middot; ' + local.counts.messages + ' messages</div>';
  h += '</div>';

  h += '<div class="gd-row" style="margin-top:14px"><button class="sec sm" id="gdCopy">Copy verdict</button><button class="sec sm" id="gdJson">Download JSON</button><button class="sec sm" id="gdAgain">Re-run</button></div>';

  box.innerHTML = h;

  $("#gdCopy").onclick = function(){
    var txt = name + " \u2014 " + score + "% gay (" + verdict + "), confidence " + conf + "%.\n" + headline + "\n" + (l2&&l2.reasoning?l2.reasoning:"");
    copy(txt, this);
  };
  $("#gdJson").onclick = function(){
    download("gaydar-" + name.replace(/\W+/g,"-").toLowerCase() + ".json", JSON.stringify({
      target:name, score:score, verdict:verdict, confidence:conf,
      layer0:state.local, layer1:state.l1, layer2:state.l2, meta:state.chat.meta
    }, null, 2));
  };
  $("#gdAgain").onclick = function(){ run(); };

  drawRadar(local);
}

function hashStr(s){ var h=0; for(var i=0;i<s.length;i++){ h=(h*31+s.charCodeAt(i))|0; } return h; }

function drawRadar(local){
  var cvs = $("#gdRadar");
  if(!cvs || !window.Chart) return;
  if(state.chart) state.chart.destroy();
  var labels = local.catList.map(function(c){ return c.label; });
  var data = local.catList.map(function(c){ return c.score; });
  state.chart = new Chart(cvs, {
    type:"radar",
    data:{ labels:labels, datasets:[{
      label:"gay-coded", data:data,
      borderColor:"#a855f7", backgroundColor:"rgba(168,85,247,0.18)",
      pointBackgroundColor:"#a855f7", pointRadius:3, borderWidth:2
    }]},
    options:{
      responsive:true, maintainAspectRatio:false,
      plugins:{ legend:{ display:false } },
      scales:{ r:{ min:0, max:100, ticks:{ display:false, stepSize:25 },
        grid:{ color:"rgba(128,128,128,0.25)" }, angleLines:{ color:"rgba(128,128,128,0.25)" },
        pointLabels:{ color:"#9aa0aa", font:{ size:10 } } } }
    }
  });
}

function copy(txt, btn){
  var done = function(){ var o=btn.textContent; btn.textContent="copied"; setTimeout(function(){ btn.textContent=o; },1400); };
  var fb = function(){
    var ta=document.createElement("textarea"); ta.value=txt; document.body.appendChild(ta); ta.select();
    try{ document.execCommand("copy"); done(); }catch(e){}
    document.body.removeChild(ta);
  };
  if(navigator.clipboard && navigator.clipboard.writeText){ navigator.clipboard.writeText(txt).then(done, fb); }
  else fb();
}

function download(fname, content){
  var b = new Blob([content], {type:"application/json"});
  var a = document.createElement("a");
  a.href = URL.createObjectURL(b); a.download = fname;
  document.body.appendChild(a); a.click();
  setTimeout(function(){ URL.revokeObjectURL(a.href); a.remove(); }, 500);
}

/* ============================== pipeline run ============================== */
function run(){
  if(state.running || !state.chat) return;
  var name = $("#gdTarget").value;
  if(!name) return;
  state.target = name; state.l2 = null; state.l1 = null;
  state.running = true;
  $("#gdRun").disabled = true;
  $("#gdPipeCard").classList.remove("gd-hidden");
  pipeErr("");
  $("#gdPipe").innerHTML = "";
  pipeStep("l0","layer 0 \u00b7 local stats","run");
  pipeStep("l1","layer 1 \u00b7 signal extraction","idle");
  pipeStep("l2","layer 2 \u00b7 verdict","idle");

  var apiKey = $("#gdKey").value.trim();
  var model = $("#gdModel").value;

  setTimeout(function(){
    state.local = layer0(name);
    pipeStep("l0","layer 0 \u00b7 local stats","done", state.local.counts.messages + " msgs");
    render();

    if(!apiKey){
      pipeStep("l1","layer 1 \u00b7 signal extraction","idle","no key");
      pipeStep("l2","layer 2 \u00b7 verdict","idle","no key");
      pipeErr("Add a Gemini API key above to run layers 1 and 2.");
      state.running = false; $("#gdRun").disabled = false;
      return;
    }

    var transcript = buildTranscript(name, 9000);
    pipeStep("l1","layer 1 \u00b7 signal extraction","run");

    runLayer1(name, transcript, model, apiKey).then(function(l1){
      state.l1 = l1;
      pipeStep("l1","layer 1 \u00b7 signal extraction","done", (l1.signals||[]).length + " signals");
      pipeStep("l2","layer 2 \u00b7 verdict","run");
      return runLayer2(name, state.local, l1, state.chat.meta, model, apiKey);
    }).then(function(l2){
      state.l2 = l2;
      pipeStep("l2","layer 2 \u00b7 verdict","done", clamp(Math.round(l2.score),0,100) + "%");
      render();
    }).catch(function(e){
      var m = String((e && e.message) || e);
      pipeStep("l2","layer 2 \u00b7 verdict","err");
      if(state.l1) pipeStep("l1","layer 1 \u00b7 signal extraction","done");
      pipeErr(m + (e && e.status===400 ? "\nTip: check the model name, or press \u201cFetch my available models\u201d." : ""));
      render();
    }).then(function(){
      state.running = false; $("#gdRun").disabled = false;
    });
  }, 30);
}

/* ============================== chat loading ============================== */
function ingest(text, sourceLabel){
  var msgs = parseWhatsApp(text);
  var real = msgs.filter(function(m){ return m.type!=="sys" && m.author!=="__sys__" && m.text; });
  if(real.length < 5){ $("#gdFileState").textContent = "Could not find messages. Is this a WhatsApp export?"; return; }

  var counts = {}, dates = [];
  real.forEach(function(m){ counts[m.author] = (counts[m.author]||0)+1; if(m.d) dates.push(m.d.getTime()); });
  var parts = Object.keys(counts).sort(function(a,b){ return counts[b]-counts[a]; });
  dates.sort(function(a,b){ return a-b; });
  var fmt = function(t){ var d=new Date(t); return d.getFullYear()+"-"+String(d.getMonth()+1).padStart(2,"0")+"-"+String(d.getDate()).padStart(2,"0"); };

  state.chat = {
    messages: msgs,
    meta: {
      participants: parts.slice(0,8).join(", ") + (parts.length>8?" +"+(parts.length-8):""),
      totalMessages: real.length,
      span: dates.length ? fmt(dates[0]) + " to " + fmt(dates[dates.length-1]) : "unknown",
      source: sourceLabel
    }
  };

  var sel = $("#gdTarget"); sel.innerHTML = "";
  parts.forEach(function(p){
    var o = document.createElement("option");
    o.value = p; o.textContent = p + "  (" + counts[p] + " messages)";
    sel.appendChild(o);
  });
  $("#gdTargetWrap").classList.remove("gd-hidden");
  $("#gdFileState").textContent = real.length.toLocaleString() + " messages parsed \u00b7 " + parts.length + " participants";
  $("#gdChatState").textContent = "Span " + state.chat.meta.span + " \u00b7 parsing happens locally, only sampled stats and excerpts are sent to Gemini.";
  $("#gdResults").innerHTML = "";
  $("#gdPipeCard").classList.add("gd-hidden");
  state.local = null; state.l1 = null; state.l2 = null;
}

function readFile(file){
  if(/\.zip$/i.test(file.name)){
    $("#gdFileState").textContent = "Reading zip\u2026";
    loadScript("https://cdnjs.cloudflare.com/ajax/libs/jszip/3.10.1/jszip.min.js")
      .then(function(){ return file.arrayBuffer(); })
      .then(function(buf){ return window.JSZip.loadAsync(buf); })
      .then(function(zip){
        var name = null;
        zip.forEach(function(p, f){ if(!f.dir && /\.txt$/i.test(p) && !name) name = p; });
        if(!name) throw new Error("No .txt found inside the zip.");
        return zip.file(name).async("string");
      })
      .then(function(txt){ ingest(txt, file.name); })
      .catch(function(e){ $("#gdFileState").textContent = "Zip failed: " + e.message + " \u2014 try the .txt export."; });
  } else {
    var r = new FileReader();
    r.onload = function(){ ingest(r.result, file.name); };
    r.onerror = function(){ $("#gdFileState").textContent = "Could not read that file."; };
    r.readAsText(file, "UTF-8");
  }
}

function loadScript(src){
  if(window.JSZip && /jszip/.test(src)) return Promise.resolve();
  return new Promise(function(res, rej){
    var s = document.createElement("script");
    s.src = src; s.onload = res; s.onerror = function(){ rej(new Error("cdn blocked")); };
    document.head.appendChild(s);
  });
}

/* ============================== demo chat ============================== */
function demoText(){
  var D = [
    ["14/03/2026","08:41","Priya","who is up. the boiler is making a noise like a dying whale"],
    ["14/03/2026","08:43","Sam","good morning my loves \u2600\uFE0F i have been awake since 5 thinking about a cushion"],
    ["14/03/2026","08:43","Sam","a cushion!!! like why"],
    ["14/03/2026","08:44","Alex","bro it is 8am"],
    ["14/03/2026","08:44","Sam","and your point is"],
    ["14/03/2026","08:45","Sam","anyway i found the most gorgeous linen one in that shop on the high street, sage green, obsessed \uD83D\uDE0D"],
    ["14/03/2026","08:47","Priya","buy the cushion sam"],
    ["14/03/2026","08:47","Sam","i already did \uD83D\uDC85"],
    ["14/03/2026","09:02","Alex","did anyone watch the match"],
    ["14/03/2026","09:03","Sam","no but i watched the highlights and honestly some of those lads could get it \uD83D\uDD25"],
    ["14/03/2026","09:03","Alex","no homo?"],
    ["14/03/2026","09:04","Sam","no homo is a cage alex"],
    ["14/03/2026","10:15","Priya","are we still doing brunch sunday"],
    ["14/03/2026","10:16","Sam","yesss obviously, i have booked the place with the good mimosas and the terrace \u2728"],
    ["14/03/2026","10:16","Priya","you are a menace and a hero"],
    ["14/03/2026","10:17","Sam","i know \uD83D\uDC85\uD83D\uDC85"],
    ["14/03/2026","11:30","Sam","okay skincare question, is the retinol before or after the serum, i keep forgetting"],
    ["14/03/2026","11:31","Priya","after. and spf every single day"],
    ["14/03/2026","11:31","Sam","i know i know, the spf is non negotiable \uD83D\uDE2D"],
    ["14/03/2026","11:32","Alex","what is a retinol"],
    ["14/03/2026","11:32","Sam","alex. sweetie. \uD83D\uDE29"],
    ["14/03/2026","13:05","Sam","rehearsal was brutal today, we are doing the whole act 2 finale and i have to belt a top B"],
    ["14/03/2026","13:06","Priya","what are you in now"],
    ["14/03/2026","13:06","Sam","community theatre production of wicked, i am a very tall and very dramatic citizen of oz"],
    ["14/03/2026","13:07","Sam","my whole cast is unhinged in the best way, we went for noodles after and someone did the whole defying gravity choreo in the street"],
    ["14/03/2026","13:08","Alex","that is so much cardio"],
    ["14/03/2026","13:09","Sam","the cardio is the point babe"],
    ["14/03/2026","15:22","Sam","look at this fit \uD83E\uDEE7 thrifted the jacket for 12 quid and i feel like a whole different person"],
    ["14/03/2026","15:23","Priya","stunning. genuinely"],
    ["14/03/2026","15:23","Sam","thank you my darling \uD83D\uDC96"],
    ["14/03/2026","16:40","Alex","gym later? leg day"],
    ["14/03/2026","16:41","Sam","yes but only if we do the good squat rack and you do not make me do that ab thing again"],
    ["14/03/2026","16:41","Alex","the ab thing is the only thing that works"],
    ["14/03/2026","16:42","Sam","it makes me feel like a dying prawn"],
    ["14/03/2026","18:02","Sam","oh my god new chappell track, i am unwell, absolutely unwell \uD83D\uDE2D\uD83D\uDE2D"],
    ["14/03/2026","18:03","Priya","send it"],
    ["14/03/2026","18:03","Sam","sent. the bridge is a religious experience"],
    ["14/03/2026","19:47","Sam","update: the cushion has arrived and the sage green is even better in person, i am rearranging the whole corner"],
    ["14/03/2026","19:48","Alex","you are rearranging a corner"],
    ["14/03/2026","19:49","Sam","a corner is a mood alex. a corner is a whole statement"],
    ["14/03/2026","21:10","Sam","right i am going out, the bar has a new cocktail list and it is all elderflower and shame"],
    ["14/03/2026","21:11","Priya","bring me back a menu"],
    ["14/03/2026","23:52","Sam","home. someone at the bar was wearing the exact jacket i thrifted, so obviously we are best friends now"],
    ["14/03/2026","23:53","Sam","we exchanged numbers and he is a dancer \uD83D\uDC83"],
    ["14/03/2026","23:54","Alex","a dancer"],
    ["14/03/2026","23:54","Sam","a dancer \uD83D\uDC83\uD83D\uDC83"],
    ["15/03/2026","00:31","Sam","anyway goodnight my loves, sweet dreams, hydrate, moisturise, be kind \u2728"],
    ["15/03/2026","00:32","Sam","and book the brunch x"],
    ["15/03/2026","07:20","Alex","morning"],
    ["15/03/2026","07:21","Sam","morning \u2600\uFE0F i have been awake since 6 thinking about the brunch menu"],
    ["15/03/2026","07:22","Alex","of course you have"],
    ["15/03/2026","07:22","Sam","babe there is a ricotta hotcake situation happening and it needs planning"],
    ["15/03/2026","09:00","Priya","sam did you buy a hat"],
    ["15/03/2026","09:01","Sam","i bought TWO hats \uD83C\uDFA9"],
    ["15/03/2026","09:01","Priya","why two"],
    ["15/03/2026","09:02","Sam","because one is for brunch and one is for the version of me who owns two hats"],
    ["15/03/2026","09:02","Sam","obviously \uD83D\uDC85"],
    ["15/03/2026","11:14","Alex","bit of a weird one but do you think that guy from the gym is fit"],
    ["15/03/2026","11:15","Sam","alex. i have been waiting EIGHT MONTHS for you to ask me that"],
    ["15/03/2026","11:15","Sam","yes. yes he is. the forearms alone. \uD83D\uDD25"],
    ["15/03/2026","11:16","Alex","ok cool cool cool"],
    ["15/03/2026","11:16","Sam","cool cool cool he says, like he has not just opened a door"],
    ["15/03/2026","14:40","Sam","rehearsal again, we are doing the flying rig and i am terrified of the flying rig"],
    ["15/03/2026","14:41","Priya","you will be amazing"],
    ["15/03/2026","14:41","Sam","if i die on that rig tell everyone i was iconic \uD83D\uDE2D"],
    ["15/03/2026","18:30","Sam","did not die. did however sing the whole drive home. the acoustics in my car are unmatched"],
    ["15/03/2026","20:05","Sam","right sunday plan: brunch 11, cushion corner photo shoot 2, nap 4, dinner 7, existential chat 9"],
    ["15/03/2026","20:06","Priya","the photo shoot"],
    ["15/03/2026","20:06","Sam","the light is best at 2 \uD83D\uDCF8"],
    ["15/03/2026","22:48","Alex","night lads"],
    ["15/03/2026","22:49","Sam","night my loves \uD83D\uDC96\uD83C\uDF08"]
  ];
  return D.map(function(r){ return "[" + r[0] + ", " + r[1] + ":00] " + r[2] + ": " + r[3]; }).join("\n");
}

/* ============================== wire up ============================== */
$("#gdDrop").addEventListener("click", function(){ $("#gdFile").click(); });
$("#gdDrop").addEventListener("dragover", function(e){ e.preventDefault(); this.classList.add("over"); });
$("#gdDrop").addEventListener("dragleave", function(){ this.classList.remove("over"); });
$("#gdDrop").addEventListener("drop", function(e){
  e.preventDefault(); this.classList.remove("over");
  if(e.dataTransfer.files && e.dataTransfer.files[0]) readFile(e.dataTransfer.files[0]);
});
$("#gdFile").addEventListener("change", function(){ if(this.files[0]) readFile(this.files[0]); });
$("#gdDemo").addEventListener("click", function(){ ingest(demoText(), "demo chat"); });
$("#gdPasteToggle").addEventListener("click", function(){ $("#gdPasteWrap").classList.toggle("gd-hidden"); });
$("#gdParsePaste").addEventListener("click", function(){ var v=$("#gdPaste").value; if(v.trim()) ingest(v, "pasted text"); });
$("#gdRun").addEventListener("click", run);

$("#gdSaveKey").addEventListener("click", function(){
  store.set("key", $("#gdKey").value.trim());
  $("#gdKeyState").textContent = "key saved locally";
  setTimeout(function(){ $("#gdKeyState").textContent = "key stored in this browser only"; }, 1600);
});

$("#gdFetchModels").addEventListener("click", function(){
  var key = $("#gdKey").value.trim(), btn = this;
  if(!key){ $("#gdKeyState").textContent = "enter a key first"; return; }
  btn.disabled = true; btn.textContent = "fetching\u2026";
  fetch("https://generativelanguage.googleapis.com/v1beta/models?key=" + encodeURIComponent(key) + "&pageSize=200")
    .then(function(r){ return r.json(); })
    .then(function(j){
      if(j.error) throw new Error(j.error.message);
      var list = (j.models||[]).filter(function(m){
        return (m.supportedGenerationMethods||[]).indexOf("generateContent")>=0 &&
          /^models\/gemini/.test(m.name) && !/(embedding|aqa|image|tts|robotics|computer-use)/.test(m.name);
      }).map(function(m){ return m.name.replace(/^models\//,""); });
      if(!list.length) throw new Error("no text models on this key");
      var sel = $("#gdModel"); sel.innerHTML = "";
      list.sort().forEach(function(n){
        var o = document.createElement("option"); o.value=n; o.textContent=n; sel.appendChild(o);
      });
      var pick = list.filter(function(n){ return /flash.*lite/i.test(n); })[0] || list.filter(function(n){ return /flash/i.test(n); })[0] || list[0];
      sel.value = pick;
      $("#gdKeyState").textContent = list.length + " models found \u00b7 using " + pick;
    })
    .catch(function(e){ $("#gdKeyState").textContent = "could not list models: " + e.message; })
    .then(function(){ btn.disabled = false; btn.textContent = "Fetch my available models"; });
});

var savedKey = store.get("key","");
if(savedKey) $("#gdKey").value = savedKey;
$("#gdKey").addEventListener("change", function(){ store.set("key", this.value.trim()); });
})();
</script>
</body>
</html>
````
